```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
print("Unique Gunnel values:", sorted(df["Gunnel"].unique()))
print("Unique Pool values:", sorted(df["Pool"].unique()))
print("Unique Water values:", sorted(df["Water"].unique()))
print("Unique Cobble values:", sorted(df["Cobble"].unique()))
print("Unique Amphiso values:", sorted(df["Amphiso"].unique()))
```

```python
#DATA TRANSFORMATION
df["Time_Hours"] = df["Time"] / 60

df[["Time", "Time_Hours"]].head()
```

```python
#time period
df["Time_Period"] = pd.cut(
    df["Time_Hours"],
    bins=[0, 6, 12, 18, 24],
    labels=["Night", "Morning", "Afternoon", "Evening"],
    include_lowest=True
)

df[["Time_Hours", "Time_Period"]].head()
```

```python
print("Gunnel values:")
print(df["Gunnel"].value_counts().sort_index())

print("\nPool values:")
print(df["Pool"].value_counts().sort_index())

print("\nWater values:")
print(df["Water"].value_counts().sort_index())

print("\nCobble values:")
print(df["Cobble"].value_counts().sort_index())

print("\nSubst values:")
print(df["Subst"].value_counts().sort_index())
```

```python
print(df[["Gunnel", "Pool", "Water", "Cobble"]].drop_duplicates().sort_values(
    ["Gunnel", "Pool", "Water", "Cobble"]
))
```

```python
df["Gunnel_Label"] = df["Gunnel"].map({
    0: "Absent",
    1: "Present"
})

df["Pool_Label"] = df["Pool"].map({
    0: "No",
    1: "Yes"
})

df["Water_Label"] = df["Water"].map({
    0: "No",
    1: "Yes"
})

df["Cobble_Label"] = df["Cobble"].map({
    0: "No",
    1: "Yes"
})
```

```python
#GUNNEL PRESENCE
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Gunnel_Label"
)

plt.title("Distribution of Gunnel Presence")
plt.xlabel("Gunnel Presence")
plt.ylabel("Number of Observations")

plt.show()
```

```python
#SHORELINE SLOPE
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Slope",
    bins=20,
    kde=True
)

plt.title("Distribution of Shoreline Slope")
plt.xlabel("Slope")
plt.ylabel("Frequency")

plt.show()
```

```python
#ALGAE COVER
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Rw",
    bins=20,
    kde=True
)

plt.title("Distribution of Rockweed/Algae Cover")
plt.xlabel("Rockweed/Algae Cover")
plt.ylabel("Frequency")

plt.show()
```

```python
#FOOD DENSITY
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Amphiso"
)

plt.title("Distribution of Crustacean Food Density")
plt.xlabel("Food Density Level")
plt.ylabel("Number of Observations")

plt.show()
```

```python
#MULTIVARIATE ANALYSIS
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Slope",
    y="Rw",
    hue="Gunnel_Label",
    size="Amphiso",
    sizes=(30, 200),
    alpha=0.7
)

plt.title(
    "Relationship Between Slope, Rockweed Cover, "
    "Food Density and Gunnel Presence"
)

plt.xlabel("Slope")
plt.ylabel("Rockweed/Algae Cover")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Time_Hours",
    y="Rw",
    hue="Gunnel_Label",
    style="Pool_Label",
    alpha=0.7
)

plt.title(
    "Time, Rockweed Cover, Gunnel Presence and Pool Condition"
)

plt.xlabel("Time of Day (Hours)")
plt.ylabel("Rockweed/Algae Cover")

plt.show()
```

```python
g = sns.FacetGrid(
    df,
    col="Pool_Label",
    hue="Gunnel_Label",
    height=5,
    aspect=1.1
)

g.map_dataframe(
    sns.scatterplot,
    x="Time_Hours",
    y="Rw",
    alpha=0.7
)

g.add_legend()

g.set_axis_labels(
    "Time of Day (Hours)",
    "Rockweed/Algae Cover"
)

g.set_titles("Pool Condition: {col_name}")

plt.show()
```

```python
plt.figure(figsize=(10, 7))

correlation_matrix = df[
    ["Gunnel", "Time", "Fromlow", "Slope", "Rw", "Amphiso"]
].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")

plt.show()
```

```python
pairplot_data = df[
    ["Gunnel", "Time", "Fromlow", "Slope", "Rw", "Amphiso"]
]

sns.pairplot(
    pairplot_data,
    hue="Gunnel",
    diag_kind="hist"
)

plt.show()
```
