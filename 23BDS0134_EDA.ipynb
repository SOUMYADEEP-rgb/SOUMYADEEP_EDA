```python
#OUTLIER DETECTION
```

```python
def iqr_outliers(series):
    Q1, Q3 = series.quantile(0.25), series.quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    return series[(series < lower) | (series > upper)]

print("Outlier counts (IQR method):\n")
for col in num_cols_1d:
    out = iqr_outliers(df[col])
    print(f"{col:10s} -> {len(out)} outliers ({len(out)/len(df)*100:.2f}%)")
```

```python
#PEARSON COORELATION
```

```python
pairs = [("Slope", "Rw"), ("Time", "Fromlow"), ("Rw", "Amphiso"), ("Slope", "Fromlow")]

print("Pearson Correlation Tests\n")
for x, y in pairs:
    r, p = pearsonr(df[x], df[y])
    sig = "Significant" if p < 0.05 else "Not Significant"
    print(f"{x} vs {y}: r={r:.3f}, p={p:.4e}  => {sig}")
```

```python
#T TEST
```

```python
print("Independent T-Tests: Gunnel Absent vs Present\n")
for col in ["Time", "Fromlow", "Slope", "Rw", "Amphiso"]:
    grp0 = df[df["Gunnel"] == 0][col]
    grp1 = df[df["Gunnel"] == 1][col]
    t_stat, p_val = ttest_ind(grp0, grp1, equal_var=False)
    sig = "Significant" if p_val < 0.05 else "Not Significant"
    print(f"{col:10s} -> t={t_stat:.3f}, p={p_val:.4e}  => {sig}")
```

```python
#CHI SQUARE TEST
```

```python
print("Chi-Square Tests of Independence with Gunnel Presence\n")
for col in ["Pool", "Water", "Cobble"]:
    contingency = pd.crosstab(df["Gunnel"], df[col])
    chi2, p, dof, expected = chi2_contingency(contingency)
    sig = "Significant" if p < 0.05 else "Not Significant"
    print(f"Gunnel vs {col:8s} -> chi2={chi2:.3f}, p={p:.4e}  => {sig}")
```

```python
#ANOVA
```

```python
groups = [df[df["Amphiso"] == level]["Rw"] for level in sorted(df["Amphiso"].unique())]
f_stat, p_val = f_oneway(*groups)
print(f"ANOVA: Rw across Amphiso levels -> F={f_stat:.3f}, p={p_val:.4e}")
print("Significant" if p_val < 0.05 else "Not Significant", "difference in Rw across food density levels")
```

```python
## 4. K-Means Clustering
```

```python
cluster_features = ["Time", "Fromlow", "Slope", "Rw", "Amphiso", "Subst"]
X_cluster = df[cluster_features].copy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)

print("Scaled data shape:", X_scaled.shape)
```

```python
wcss = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, init="k-means++", random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, wcss, marker='o', color="darkorange")
plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS (Inertia)")
plt.xticks(K_range)
plt.show()
```

```python
sil_scores = []
K_range_sil = range(2, 11)

for k in K_range_sil:
    kmeans = KMeans(n_clusters=k, init="k-means++", random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    sil_scores.append(score)
    print(f"k={k}: Silhouette Score = {score:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(list(K_range_sil), sil_scores, marker='o', color="seagreen")
plt.title("Silhouette Scores for Different k")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

```python
optimal_k = 4

kmeans_final = KMeans(n_clusters=optimal_k, init="k-means++", random_state=42, n_init=10)
df["KMeans_Cluster"] = kmeans_final.fit_predict(X_scaled)

print(df["KMeans_Cluster"].value_counts().sort_index())
```

```python
pca = PCA(n_components=2)
pca_components = pca.fit_transform(X_scaled)

df["PCA1"] = pca_components[:, 0]
df["PCA2"] = pca_components[:, 1]

plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x="PCA1", y="PCA2", hue="KMeans_Cluster", palette="Set2", alpha=0.75)
plt.title("K-Means Clusters Visualized via PCA (2 Components)")
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% variance)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% variance)")
plt.show()
```

```python
cluster_profile = df.groupby("KMeans_Cluster")[cluster_features].mean()
print("K-Means Cluster Profiles (mean values):\n")
print(cluster_profile)
```
